```python
import pandas as pd
import matplotlib.pyplot as plt
import gradio as gr

# === Sample dataset ===
roles_present = [
    {"Role": "Software Engineer", "Skills": "Python, Databases, APIs"},
    {"Role": "Data Analyst", "Skills": "SQL, Excel, Visualization"},
    {"Role": "Cybersecurity Analyst", "Skills": "Networking, Security Tools"}
]

roles_future = [
    {"Role": "AI Ethics Officer", "Skills": "AI Policy, Ethics, Law"},
    {"Role": "Climate Data Scientist", "Skills": "AI, Climate Models, Statistics"},
    {"Role": "Neural Interface Designer", "Skills": "Neuroscience, ML, UX"}
]

df_roles_present = pd.DataFrame(roles_present)
df_roles_future = pd.DataFrame(roles_future)

# === Prepare skills list for dropdown ===
all_skills = set()
for skills in pd.concat([df_roles_present["Skills"], df_roles_future["Skills"]]):
    all_skills.update([s.strip() for s in skills.split(",")])
all_skills = sorted(list(all_skills))

# === Helper: save plot ===
def plot_jobs(df, title, filename):
    plt.figure(figsize=(6,4))
    plt.barh(df["Role"], range(len(df)), color="skyblue")
    plt.title(title)
    plt.tight_layout()
    plt.savefig(filename)
    plt.close()
    return filename

# === Main function with dynamic comparison ===
def recommend_roles(skills):
    if not skills:
        skills = []  # fallback to empty list
    skills = [s.strip().lower() for s in skills]

    # Filter present jobs
    present_jobs = df_roles_present[df_roles_present["Skills"].apply(
        lambda x: any(skill in x.lower() for skill in skills)
    )]

    # Filter future jobs
    future_jobs = df_roles_future[df_roles_future["Skills"].apply(
        lambda x: any(skill in x.lower() for skill in skills)
    )]

    # If nothing matches, show all (fallback)
    if present_jobs.empty:
        present_jobs = df_roles_present
    if future_jobs.empty:
        future_jobs = df_roles_future

    # === Dynamic, scenario-based comparison ===
    def generate_comparison(skills, present_jobs, future_jobs):
        skills_set = set([s.lower() for s in skills])

        # Present jobs skills
        present_skills = set()
        for s in present_jobs["Skills"]:
            present_skills.update([x.strip().lower() for x in s.split(",")])

        # Future jobs skills
        future_skills = set()
        for s in future_jobs["Skills"]:
            future_skills.update([x.strip().lower() for x in s.split(",")])

        # Skills match
        present_match = skills_set & present_skills
        future_gap = future_skills - skills_set

        comparison = "### Career Comparison Report\n\n"

        if present_jobs.empty:
            comparison += "⚠️ No present jobs match your current skills.\n"
        else:
            comparison += f"✅ Present jobs match these skills: **{', '.join(present_match)}**\n"
            comparison += f"   You are well suited for: **{', '.join(present_jobs['Role'])}**\n"

        if future_jobs.empty:
            comparison += "⚠️ No future jobs found for your skill set.\n"
        else:
            comparison += f"🔮 Future jobs may require additional skills: **{', '.join(future_gap)}**\n"
            comparison += f"   Consider upskilling for roles like: **{', '.join(future_jobs['Role'])}**\n"

        return comparison

    comparison_text = generate_comparison(skills, present_jobs, future_jobs)

    # Roadmap
    roadmap_text = "### Roadmap\n\nStart with your skills → Expand into Data/AI → Add specialized future skills"

    # Save charts
    present_chart_file = plot_jobs(present_jobs, "Present Careers", "present.png")
    future_chart_file = plot_jobs(future_jobs, "Future Careers", "future.png")

    # Save CSV
    csv_path = "future_jobs.csv"
    future_jobs.to_csv(csv_path, index=False)

    return present_jobs, future_jobs, comparison_text, roadmap_text, present_chart_file, future_chart_file, csv_path

# === Combined input function for dropdown + manual entry ===
def combined_recommend_roles(dropdown_skills, textbox_skills):
    # Convert textbox input to list
    if not textbox_skills:
        manual_skills = []
    else:
        manual_skills = [s.strip().lower() for s in textbox_skills.split(",")]

    # Dropdown skills
    dropdown_skills = [s.strip().lower() for s in dropdown_skills]

    # Combine both
    all_selected_skills = dropdown_skills + manual_skills

    return recommend_roles(all_selected_skills)

# === Gradio UI ===
with gr.Blocks() as demo:
    with gr.Row():
        skills_dropdown = gr.Dropdown(
            label="Select your skills",
            choices=all_skills,
            value=[],  # default empty
            multiselect=True
        )
        skills_textbox = gr.Textbox(
            label="Or enter your own skills (comma separated)",
            placeholder="e.g., Python, SQL, Networking"
        )
        btn = gr.Button("Find Careers")

    present_table = gr.Dataframe()
    future_table = gr.Dataframe()
    comparison_md = gr.Markdown()
    roadmap_md = gr.Markdown()
    present_chart_img = gr.Image(type="filepath")
    future_chart_img = gr.Image(type="filepath")
    file_out = gr.File(label="Download Future Jobs CSV")

    btn.click(
        fn=combined_recommend_roles,
        inputs=[skills_dropdown, skills_textbox],
        outputs=[present_table, future_table, comparison_md, roadmap_md, present_chart_img, future_chart_img, file_out]
    )

demo.launch()
```
